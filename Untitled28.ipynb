```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('Boston.csv')
df.head()
```

```python
plt.hist(df['nox'], bins=10)
plt.xlabel('Nox Levels')
plt.ylabel('Frequency')
plt.title('Histogram of nOx Levels')
plt.show()
```

```python
X = pd.DataFrame(df.iloc[:, :-1])
Y = pd.DataFrame(df.iloc[:, -1])
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X ,Y, test_size=0.2)
```

```python
from sklearn.ensemble import RandamForestRegressor
regressor = RandamForestRegressor(n_estimators=20, random_state=0)
regressor.fit(X_train, Y_train)
Y_pred = regressor.predict(X_test)
```

```python
from sklearn import metrics

print("Mean Absolute Error:",metrics.mean_absolute_error(Y_test, Y_pred))
print("Mean squared Error:",metrics.mean_squared_error(Y_test, Y_pred))
print("Root Mean Squared Error:", np.sqrt(metrics.mean_sqared_error(Y_test)))
```
